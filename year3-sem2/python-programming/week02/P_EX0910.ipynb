# 함수
- print() : 값을 화면에 출력하는 함수
- type() : 값이나 변수의 자료형을 확인하는 함수
- print(type()) : 값이나 변수의 자료형을 화면에 출력할 때 사용
- input() : 사용자로부터 값을 입력받는 함수
- int() : 값을 정수형(int)으로 변환하는 함수

```python
n1 = input("첫번째 정수 입력: ")    # 2
n2 = input("두번째 정수 입력: ")    # 4
result = n1 + n2       # 문자열 덧셈 연산 : 문자열 연결
print(result)

n3 = int(input("첫번째 정수 입력: "))    # 2
n4 = int(input("두번째 정수 입력: "))    # 4
sum = n3 + n4       # 정수형 덧셈 연산 : 계산
print(n1, "과", n2, "의 합은", sum, "입니다.")
```
